```python
import requests
from langchain_core.tools import tool
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
```

```python
def getFlights(origin,ziel,datum_bis):
    res = requests.post(
        "https://mcp.octotrip.app/flights/mcp",
        headers={
            "Content-Type":"application/json",
            "Accept":"application/json, text/event-stream",
        },
        json={
            "jsonrpc": "2.0",
  "id": 2,
  "method": "tools/call",
  "params": {
    "name": "search",
    "arguments": {
      "origin": origin,
      "destination": ziel,
      "departure_date": datum_bis
    }
  }
        }
    )
    res.raise_for_status()
    return res
```

```python
response = getFlights("DUS","TUN","2026-11-16")
print(response.text)
```

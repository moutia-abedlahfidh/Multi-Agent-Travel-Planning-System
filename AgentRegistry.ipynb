```python
class AgentRegistry:
    def __init__(self):
        self.agents = {}
        self.beschreibungen = {}

    def register(self, name, beschreibung, agent):
        self.agents[name] = agent
        self.beschreibungen[name] = beschreibung

    def list_beschreibungen(self):
        return self.beschreibungen

    def list_agents(self):
        return self.agents

    def call(self,name,anfrage):
        res = self.agents[name].invoke(
        {"messages":[{"role":"user","content":anfrage}]}
    )
        return res["messages"][-1].content
```

```python
registry = AgentRegistry()
```

```python
registry.register("Flight_Agent", "Flüge zwischen IATA-Codes, Preise, Zeiten", flight_agent)
```

```python
registry.register("Weather_Agent", "Wetter für Ort und Datum", weather_agent)
```

```python
print(registry.list_agents())
```

```python
print(registry.list_beschreibungen())
```

```python
print(registry.call("Flight_Agent","Flights from Köln nach Tunesien am 21.11.2026"))
```
